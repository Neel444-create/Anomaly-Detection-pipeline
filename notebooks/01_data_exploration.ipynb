# Credit Card Fraud Data Exploration

This notebook reads the raw dataset without changing it. Reusable exploration logic lives in `src/data/explore.py`; this notebook focuses on visual inspection.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

# Works when launched from either the project root or notebooks/.
project_root = Path.cwd()
if not (project_root / "data" / "raw" / "creditcard.csv").is_file():
    project_root = project_root.parent

data_path = project_root / "data" / "raw" / "creditcard.csv"
df = pd.read_csv(data_path)
df.shape
```

```python
# Class distribution: 0 is legitimate and 1 is fraudulent.
class_counts = df["Class"].value_counts().sort_index()
ax = class_counts.plot(kind="bar", color=["#4C78A8", "#E45756"], rot=0)
ax.set(
    title="Transaction Class Distribution",
    xlabel="Class",
    ylabel="Number of transactions",
)
for bar, count in zip(ax.patches, class_counts):
    ax.annotate(
        f"{count:,}",
        (bar.get_x() + bar.get_width() / 2, count),
        ha="center",
        va="bottom",
        fontsize=9,
    )
plt.tight_layout()
```

```python
# A 99th-percentile cap keeps a few extreme amounts from hiding the main distribution.
amount_cap = df["Amount"].quantile(0.99)
ax = (
    df["Amount"]
    .clip(upper=amount_cap)
    .plot(kind="hist", bins=60, figsize=(9, 4), color="#4C78A8", edgecolor="white")
)
ax.set(
    title="Transaction Amount Distribution (capped at 99th percentile)",
    xlabel="Amount",
    ylabel="Number of transactions",
)
plt.tight_layout()
```

```python
# Basic numerical statistics for every numerical feature.
feature_statistics = df.describe().T
display(feature_statistics)

# Visualize feature standard deviations; the target is excluded because it is binary.
feature_std = feature_statistics.drop(index="Class")["std"].sort_values()
ax = feature_std.plot(kind="barh", figsize=(8, 8), color="#72B7B2")
ax.set(
    title="Feature Standard Deviations", xlabel="Standard deviation", ylabel="Feature"
)
plt.tight_layout()
```
